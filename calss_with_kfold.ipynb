## K-FOLD
Ele divide os dados em K partes e treina/testa o modelo várias vezes, usando uma parte diferente para teste em cada vez.

```python
from sklearn.model_selection import KFold
from sklearn.naive_bayes import GaussianNB
```

```python
data = load_breast_cancer()
x = data.data
y = data.target
```

```python
gnb = GaussianNB()
```

```python
import numpy as np

kf = KFold(n_splits=5)

Acc = []
f1_Score = []


heat_map(data,gnb)
for i, (train_index, test_index) in enumerate(kf.split(x)):
    gnb.fit(x[train_index], y[train_index])
    y_pred = gnb.predict(x[test_index])
    Acc.append(accuracy_score(y[test_index], y_pred))
    f1_Score.append(f1_score(y[test_index], y_pred))

print(f"Acuracia geral = {np.mean(Acc)}")
print(f"F1_score Geral = {np.mean(f1_Score)}")
```
