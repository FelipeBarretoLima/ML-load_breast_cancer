```python
from sklearn.datasets import load_diabetes
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = load_diabetes()
x = data.data
y = data.target
```

```python
print(data.DESCR)
```

```python
print(data.feature_names)
```

```python
from sklearn.metrics import root_mean_squared_error, mean_squared_error, r2_score
from sklearn.model_selection import KFold
import numpy as np
```

```python
from sklearn import linear_model
reg = linear_model.LinearRegression()

mse = []
rmse = []
r2 = []
b0 = []
b1 = []


kf = KFold(n_splits=5)

for i, (train_index, test_index) in enumerate(kf.split(x)):
    reg.fit(x[train_index], y[train_index])
    y_pred = reg.predict(x[test_index])
    mse.append(mean_squared_error(y[test_index], y_pred))
    rmse.append(root_mean_squared_error(y[test_index], y_pred))
    r2.append(r2_score(y[test_index],y_pred))
    b0.append(reg.intercept_)
    b1.append(reg.coef_)

b_medias = np.mean(b1, axis=0)

print(f"Erro quadrático médio = {np.mean(mse)}")
print(f"Raiz do Erro quadrático médio = {np.mean(rmse)}")
print(f"R² = {np.mean(r2)}")
print(f"b0 = {np.mean(b0):.2f}")

for i, coef in enumerate(b_medias, start=1):
    print(f"b{i} = {coef:.4f}", end=" | ")
```
